```python
import pandas as pd
```

```python
try:
  df = pd.read_csv('sample_data.csv',parse_dates=['order_date'],dtype={
      'payment_method': 'category'
  })
  print(df.info())
  print(df.describe())
except:
  print('Error reading file')

```

```python
print(df['revenue'].quantile([0.25,0.5,0.75]))
```

```python
print(df['revenue'].mean())
print(df['revenue'].median())
```

```python
print(df['revenue'].skew())
```

```python
desc = df.groupby('product_category')['revenue'].describe()
```

```python
total_sum = df.groupby('product_category')['revenue'].sum()
desc.insert(1,'sum',total_sum)

desc
```

```python
df.groupby('product_category')['revenue'].mean()
```

```python
Q1 = df['revenue'].quantile(0.25)
Q3 = df['revenue'].quantile(0.75)

IQR = Q3 - Q1
print(IQR, Q1,Q3)

df = df.set_index('order_date')
```

```python

monthly_revenue = df['revenue'].resample('YE').sum()
monthly_revenue.pct_change() * 100
```

```python
Q1 = df['revenue'].quantile(0.25)
Q3 = df['revenue'].quantile(0.75)
IQR = Q3 - Q1
print(IQR,Q3,Q1)
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
print(lower,upper)
# outliers_iqr = df[(df['revenue'] < lower) | (df['revenue'] > upper)]
# outliers_iqr.shape

df = df[(df['revenue'] > lower) & (df['revenue'] < upper)]
df.shape
```

```python
df.info()
```

```python
numerical_df = df[['quantity','unit_price','discount','delivery_days','customer_rating','revenue']]
print(numerical_df.corr())
```
