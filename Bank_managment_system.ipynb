```python
class LessAmountException(Exception):
  pass


class BankAccount:
  def __init__(self,owner,account_number,account_balance=0):
    self.owner = owner
    self.__account_number = account_number
    self.__account_balance = account_balance
    self.transactions = []
  def deposit(self,amount):
    try:
      if(amount > 0):
        self.account_balance+=amount
        self.transactions.append(f"Deposited ${amount}")
    except LessAmountException as e:
      print(f"Error in amount you want to deposit : {e}")
  def withdraw(self,amount):
    try:
      if(amount > 0 and  amount <= self.account_balance):
        self.account_balance-=amount
        self.transactions.append(f"Withdrew ${amount}")
    except LessAmountException as e:
      print(f"Error in amount you want to withdraw : {e}")

  @property
  def account_balance(self):
    return self.__account_balance
  @account_balance.setter
  def account_balance(self,amount):
    try:
      if(amount<0):
        raise LessAmountException("Amount can't be negative")
      self.__account_balance = amount
    except LessAmountException as e:
      print(f"Error in amount you want to deposit : {e}")



  def print_history(self):
    print(f' Bank account owner name is: {self.owner}\nAccount balance right now is{self.account_balance}\nAccount number is {self.__account_number}')
    print("Following is the transaction history")
    for transaction in self.transactions:
      print(transaction)

  def __str__(self):
    return f' Bank account owner name is: {self.owner}\nAccount balance right now is{self.account_balance}\nAccount number is {self.__account_number}'

  def make_transaction_statement(self,):
    textfile_name = self.owner.replace(" ","_")
    try:
      with open(f"{textfile_name}.txt","w") as file:
        file.write(f' Bank account owner name is: {self.owner}\nAccount balance right now is{self.account_balance}\nAccount number is {self.__account_number}\n')
        for transaction in self.transactions:
          file.write(transaction)
          file.write("\n")
    except:
      print("Error in creating file")

Sarim_Acc = BankAccount('Sarim',182637,20000)
Sarim_Acc.deposit(10000)
Sarim_Acc.withdraw(5000)
Sarim_Acc.print_history()
Sarim_Acc.make_transaction_statement()









```
