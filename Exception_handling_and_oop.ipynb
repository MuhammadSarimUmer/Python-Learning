```python
class Car():


  def __init__(self,colour):
    self.colour = colour

  def myColour(self):
    print(f'my colour is {self.colour}')

  def myColour2(self):
    colour2 = 'blue'
    print(f'my colour is {colour2}')

  def changeColour(self,newColour):

    self.colour = newColour
    print(f'my new colour is {self.colour}')

car1 = Car('red')

car1.myColour()
car1.myColour2()
car1.changeColour('green')
```

```python
class Animal:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def myName(self):
    print(f'my name is {self.name}')

  def myAge(self):
    print(f'my age is {self.age}')

cat = Animal('cat',5)

cat.myName()
cat.myAge()

cat.age = 7

cat.myAge()
```

```python
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def studentInfo(self):
        print(f"Student name is {self.name}")
        print(f"Student age is {self.age}")
        print(f"Student grade is {self.grade}")

    def inputInfo(self):
        self.name = input("Enter student name: ")
        self.age = int(input("Enter student age: "))
        self.grade = input("Enter student grade: ")

    def eligibilityCriteria(self):
        if self.age > 18:
            print(f"{self.name} is eligible for admission in University")
        else:
            print(f"{self.name} is not eligible")


Sarim = Student('Sarim', 19, 'A+')
Sarim.studentInfo()
Sarim.eligibilityCriteria()
Sarim.inputInfo()
Sarim.studentInfo()
Sarim.eligibilityCriteria()
```

```python
class Student:
  def __init__(self,grade):
    self.__grade = grade

  def getGrade(self):
    return self.__grade

sarim = Student('C')

print(sarim.getGrade())
```

```python
class Student:
  def __init__(self,grade):
    self.__grade = grade

  @property
  def grade(self):
    return self.__grade

  @grade.setter
  def grade(self,value):
    if value not in ['A','B','C','D','E','F']:
      print('Invalid grade')
    else:
      self.__grade = value

sarim = Student('A')

print(sarim.grade)

sarim.grade = 'B'

print(sarim.grade)
```

```python

try:
  number = int(input('Enter a Number: '))
  print(10/number)
except ZeroDivisionError:
  print('You cannot divide by 0')
except ValueError:
  print('Invalid input')
except Exception as e:
  print(f'an unknown error occured: {e}')

class InvalidAgeError(Exception):
  pass


age = -9
if age<0:
  raise InvalidAgeError('Age cannot be negative')
else:
  print(f'AGE IS {age}')
```
