```python
salary = float(input("Enter salary: "))

if salary < 30000:
    tax = salary * 0.05
elif salary <= 70000:
    tax = salary * 0.15
else:
    tax = salary * 0.25

print("Final Tax:", tax)
```

```python
class User:
    def __init__(self, name):
        self.name = name

class Message:
    def __init__(self, sender, text):
        self.sender = sender
        self.text = text

class ChatRoom:
    def __init__(self):
        self.users = []
        self.history = []

    def join(self, user):
        self.users.append(user)
        print(user.name, "joined the room.")

    def leave(self, user):
        self.users.remove(user)
        print(user.name, "left the room.")

    def send_message(self, user, text):
        msg = Message(user.name, text)
        self.history.append(msg)

    def view_history(self):
        for msg in self.history:
            print(msg.sender + ": " + msg.text)

room = ChatRoom()
u1 = User("Sarim")
u2 = User("Muhammad")

room.join(u1)
room.join(u2)
room.send_message(u1, "Salam")
room.send_message(u2, "Walaikum salam")
room.view_history()
room.leave(u1)
```
