```python
# make a nested dictiory to store rollnumbers,name and scores and  gpa
```

```python
students = {}

def add_students():
  try:

    student_roll = input('Enter roll number: ')
    if not student_roll.isdigit():
      print('Invalid roll number')
      return
    student_name = input('Enter name: ')
    if not student_name.strip():
      print('Invalid name')
      return
    students[student_roll] = {
    'name': student_name,
    'scores': []
     }
    print(f'Student {student_name} is added with roll number {student_roll}')
  except Exception as e:
    print(f'An error occured : {e}')
    return
def add_scores(student_roll):
  try:
    student_scores  = []
    print('Enter numbers upto 5 subjects: \n')
    for score in range(5):
      print(f'Enter score for subject number {score + 1}')
      input_roll = input()
      if not input_roll.isdigit():
        print('Invalid score')
        return
      student_scores.append(int(input_roll))
    students[student_roll]['scores'] = student_scores
  except Exception as e:
    print(f'An error occured : {e}')
    return

def calculate_gpa(student_roll = None):
  try:
    if(student_roll == None):
      print('Student Roll not found or given, checking the name')
    else:
      total = sum(students[student_roll]['scores'])
      gpa = total/5
      return gpa
  except Exception as e:
    print(f'An error occured : {e}')
    return
def print_report():
  with open('report.txt', 'w') as file:
    for student_roll, student_data in students.items():
      student_name = student_data['name']
      student_scores = student_data['scores']
      student_gpa = calculate_gpa(student_roll)
      file.write(f'{student_roll} {student_name} {student_scores} {student_gpa}\n')
      print(f'{student_roll} {student_name} {student_scores} {student_gpa}')



add_students()
add_scores('35')
calculate_gpa('35')
print_report()
```
