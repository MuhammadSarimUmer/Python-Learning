```python
import random as rd

print('Welcome to the Number guessing game!!')
print('Please select your difficulty')

while True:
    try:

        user_input = int(input('1. Easy(guess from 1-5)\n2. Medium(guess from 1-10)\n3. Hard(guess from 1-20)\n'))

        if user_input == 1:
            random_num = rd.randint(1, 5)
            break
        elif user_input == 2:
            random_num = rd.randint(1, 10)
            break
        elif user_input == 3:
            random_num = rd.randint(1, 20)
            break
        else:
            print('Please enter 1, 2, or 3.')
    except ValueError:
        print('Wrong input!! Please enter a number.')

for attempts in range(3):
    print(f"Attempt number {attempts + 1}")

    try:

        last_input = int(input('Enter your number(should be positive): '))
    except ValueError:
        print('Wrong input!! Please enter a valid number.')
        continue

    if last_input < 0:
        print('Error, your number is not positive')
        continue

    if last_input == random_num:
        print('You won!!')
        break
    elif last_input > random_num:
        print('You lost this attempt!! (Too high)')
    elif last_input < random_num:
        print('You lost this attempt!! (Too low)')
```
